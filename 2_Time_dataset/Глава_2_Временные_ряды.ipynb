```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.tsa.seasonal import seasonal_decompose
import os
import warnings
from scipy.stats import skew, kurtosis, norm

warnings.filterwarnings('ignore')

# Настройка стиля графиков
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# =============================================================================
# КЛАСС ДЛЯ АНАЛИЗА ВРЕМЕННЫХ РЯДОВ SKAB
# =============================================================================

class SKABAnomalyAnalyzer:
    """УММ по анализу временных рядов"""

    def __init__(self, data_path="/content", critical_temp=85):
        self.data_path = data_path
        self.critical_temp = critical_temp
        self.data = None
        self.anomalies = None
        self.file_info = []

    def load_all_data(self):
        """Загрузка и первичное знакомство с данными"""
        print("=" * 70)
        print("ЭТАП 1: ЗАГРУЗКА И ПЕРВИЧНОЕ ЗНАКОМСТВО С ДАННЫМИ")
        print("=" * 70)

        if not os.path.exists(self.data_path):
            raise FileNotFoundError(f"Папка {self.data_path} не найдена!")

        # Поиск всех CSV-файлов в папке
        csv_files = [f for f in os.listdir(self.data_path)
                    if f.endswith('.csv') and os.path.isfile(os.path.join(self.data_path, f))]

        print(f"Найдено CSV файлов: {len(csv_files)}")

        if len(csv_files) == 0:
            raise FileNotFoundError(f"В папке {self.data_path} нет CSV-файлов!")

        all_dfs = []
        total_rows = 0

        for idx, csv_file in enumerate(sorted(csv_files), 1):
            file_path = os.path.join(self.data_path, csv_file)

            try:
                # Определение разделителя (точка с запятой для SKAB)
                with open(file_path, 'r', encoding='utf-8') as f:
                    first_line = f.readline()
                    delimiter = ';' if ';' in first_line else ','

                # Загрузка данных с оптимизацией типов
                df = pd.read_csv(file_path, delimiter=delimiter,
                                 dtype={'Accelerometer1RMS': 'float32',
                                        'Accelerometer2RMS': 'float32',
                                        'Current': 'float32',
                                        'Pressure': 'float32',
                                        'Temperature': 'float32',
                                        'Thermocouple': 'float32',
                                        'Voltage': 'float32',
                                        'Volume Flow RateRMS': 'float32',
                                        'anomaly': 'int8',
                                        'changepoint': 'int8'})

                # Преобразование временной метки
                df['datetime'] = pd.to_datetime(df['datetime'])

                # Добавление информации о файле
                df['file_id'] = idx
                df['file_name'] = csv_file

                rows_count = len(df)
                total_rows += rows_count

                self.file_info.append({
                    'file_id': idx,
                    'file_name': csv_file,
                    'rows': rows_count,
                    'anomaly_count': df['anomaly'].sum() if 'anomaly' in df.columns else 0
                })

                all_dfs.append(df)
                print(f"  [{idx}/{len(csv_files)}] Загружен {csv_file}: {rows_count} записей, "
                      f"аномалий: {df['anomaly'].sum() if 'anomaly' in df.columns else 0}")

            except Exception as e:
                print(f"  Ошибка при загрузке {csv_file}: {e}")
                continue

        # Объединение всех данных
        self.data = pd.concat(all_dfs, ignore_index=True)

        # Создание временной метки-индекса
        self.data['timestamp'] = range(len(self.data))

        # Сохранение порядка файлов для дальнейшего анализа
        self.data['file_id'] = self.data['file_id'].astype('category')

        print("\n" + "-" * 70)
        print("РЕЗУЛЬТАТЫ ЗАГРУЗКИ:")
        print(f"Общее количество записей: {len(self.data):,}")
        print(f"Количество файлов: {len(csv_files)}")
        print(f"Каналы (измерения): 8 сенсорных каналов + 2 метки аномалий")
        print(f"Временной диапазон: с {self.data['datetime'].min()} по {self.data['datetime'].max()}")
        print(f"Наличие аномалий: {self.data['anomaly'].sum() if 'anomaly' in self.data.columns else 0} записей")

        # Вывод информации о типах данных
        print("\nТипы данных в каналах:")
        numeric_cols = ['Accelerometer1RMS', 'Accelerometer2RMS', 'Current',
                       'Pressure', 'Temperature', 'Thermocouple', 'Voltage', 'Volume Flow RateRMS']
        for col in numeric_cols:
            if col in self.data.columns:
                print(f"{col}: {self.data[col].dtype}")

        import gc
        gc.collect()

    def visualize_initial_data(self):
        """Визуализация исходных данных"""
        print("\n" + "=" * 70)
        print("ЭТАП 2: ВИЗУАЛИЗАЦИЯ ИСХОДНЫХ ДАННЫХ")
        print("=" * 70)

        if self.data is None or len(self.data) == 0:
            print("  Нет данных для визуализации")
            return

        # Даунсемплинг для визуализации (если данных слишком много)
        total_points = len(self.data)
        max_points_per_plot = 100000

        if total_points > max_points_per_plot:
            step = total_points // max_points_per_plot
            viz_data = self.data.iloc[::step].copy()
            print(f"Даунсемплинг: {total_points:,} → {len(viz_data):,} точек для визуализации")
        else:
            viz_data = self.data

        # Определение каналов для визуализации
        channels = [
            ('Accelerometer1RMS', 'Вибрация (акселерометр 1)', 'Амплитуда (g)', 'blue'),
            ('Accelerometer2RMS', 'Вибрация (акселерометр 2)', 'Амплитуда (g)', 'green'),
            ('Current', 'Ток', 'Сила тока (А)', 'orange'),
            ('Pressure', 'Давление', 'Давление (Бар)', 'red'),
            ('Temperature', 'Температура корпуса', 'Температура (°C)', 'darkred'),
            ('Thermocouple', 'Температура жидкости', 'Температура (°C)', 'purple'),
            ('Voltage', 'Напряжение', 'Напряжение (В)', 'cyan'),
            ('Volume Flow RateRMS', 'Расход жидкости', 'Расход (л/мин)', 'brown')
        ]

        fig, axes = plt.subplots(4, 2, figsize=(16, 14))
        axes = axes.flatten()

        for idx, (col, title, ylabel, color) in enumerate(channels):
            if col not in viz_data.columns:
                axes[idx].set_visible(False)
                continue

            # Линейный график
            axes[idx].plot(viz_data['timestamp'], viz_data[col],
                          color=color, linewidth=0.6, alpha=0.7, rasterized=True)

            # Выделение аномалий (если есть)
            if 'anomaly' in viz_data.columns:
                anomaly_mask = viz_data['anomaly'] == 1
                if anomaly_mask.any():
                    axes[idx].scatter(viz_data[anomaly_mask]['timestamp'],
                                     viz_data[anomaly_mask][col],
                                     color='red', s=15, alpha=0.8,
                                     label='Аномалии', zorder=5)

            # Статистические линии
            mean_val = viz_data[col].mean()
            std_val = viz_data[col].std()
            axes[idx].axhline(y=mean_val, color='black', linestyle='--',
                            linewidth=1, alpha=0.5, label=f'Среднее: {mean_val:.2f}')
            axes[idx].axhline(y=mean_val + 3*std_val, color='gray',
                            linestyle=':', linewidth=0.8, alpha=0.5)
            axes[idx].axhline(y=mean_val - 3*std_val, color='gray',
                            linestyle=':', linewidth=0.8, alpha=0.5)

            axes[idx].set_ylabel(ylabel, fontsize=10)
            axes[idx].set_title(title, fontsize=11, fontweight='bold')
            axes[idx].grid(True, alpha=0.3)
            axes[idx].legend(fontsize=8, loc='upper right')

        # Скрыть неиспользуемые подграфики
        for idx in range(len(channels), len(axes)):
            axes[idx].set_visible(False)

        axes[-2].set_xlabel('Временные точки (порядковый номер)', fontsize=10)
        axes[-1].set_xlabel('Временные точки (порядковый номер)', fontsize=10)

        plt.suptitle('Многомерный временной ряд SKAB: все сенсорные каналы\n'
                    '(красные точки — размеченные аномалии)',
                    fontsize=14, fontweight='bold', y=1.02)
        plt.tight_layout()
        plt.savefig('01_initial_time_series.png', dpi=150, bbox_inches='tight')
        plt.show()

        print("\nВыводы по этапу 2:")
        print("Графики показывают поведение всех 8 сенсорных каналов")
        print("Аномалии (красные точки) визуально выделяются на графиках")
        print("Наблюдаются различные типы аномалий: точечные и коллективные")
        print("Разрывы на графиках отсутствуют → нет пропусков")

    def statistical_analysis(self):
        """Статистический анализ"""
        print("\n" + "=" * 70)
        print("ЭТАП 3: СТАТИСТИЧЕСКИЙ АНАЛИЗ")
        print("=" * 70)

        if self.data is None:
            print("Нет данных для анализа")
            return

        channels = ['Accelerometer1RMS', 'Accelerometer2RMS', 'Current',
                   'Pressure', 'Temperature', 'Thermocouple', 'Voltage', 'Volume Flow RateRMS']
        existing_channels = [ch for ch in channels if ch in self.data.columns]

        # Расчет статистик
        stats_df = pd.DataFrame()
        for col in existing_channels:
            stats_df.loc['count', col] = self.data[col].count()
            stats_df.loc['mean', col] = self.data[col].mean()
            stats_df.loc['std', col] = self.data[col].std()
            stats_df.loc['min', col] = self.data[col].min()
            stats_df.loc['Q1 (25%)', col] = self.data[col].quantile(0.25)
            stats_df.loc['median (Q2)', col] = self.data[col].median()
            stats_df.loc['Q3 (75%)', col] = self.data[col].quantile(0.75)
            stats_df.loc['max', col] = self.data[col].max()

        print("\nТаблица 1. Статистические характеристики каналов:")
        print(stats_df.round(4).to_string())

        # Анализ симметричности распределений
        print("\nАнализ распределения (сравнение среднего и медианы):")
        for col in existing_channels:
            mean_val = self.data[col].mean()
            median_val = self.data[col].median()
            if mean_val != 0:
                diff_pct = abs(mean_val - median_val) / mean_val * 100
            else:
                diff_pct = 0
            symmetry = "симметричное" if diff_pct < 5 else "асимметричное"
            print(f"{col}: среднее={mean_val:.4f}, медиана={median_val:.4f} → {symmetry} "
                  f"(отклонение {diff_pct:.1f}%)")

        # Определение частоты дискретизации
        if 'datetime' in self.data.columns:
            time_diffs = self.data['datetime'].diff().dropna()
            if len(time_diffs) > 0:
                most_freq = time_diffs.mode().iloc[0]
                print(f"\nЧастота дискретизации: {most_freq} (секундная)")

        print(f"\nРазброс значений (std):")
        for col in existing_channels:
            print(f"      - {col}: {self.data[col].std():.4f}")

        # Визуализация распределений
        fig, axes = plt.subplots(2, 4, figsize=(16, 10))
        axes = axes.flatten()

        for idx, col in enumerate(existing_channels):
            self.data[col].hist(bins=100, ax=axes[idx], alpha=0.7,
                               edgecolor='black', linewidth=0.5)
            axes[idx].set_title(f'Распределение: {col}', fontsize=10, fontweight='bold')
            axes[idx].set_xlabel('Значение')
            axes[idx].set_ylabel('Частота')
            axes[idx].axvline(self.data[col].mean(), color='red', linestyle='--',
                            linewidth=1.5, label=f'Среднее: {self.data[col].mean():.2f}')
            axes[idx].axvline(self.data[col].median(), color='green', linestyle='--',
                            linewidth=1.5, label=f'Медиана: {self.data[col].median():.2f}')
            axes[idx].legend(fontsize=8)
            axes[idx].grid(True, alpha=0.3)

        for idx in range(len(existing_channels), len(axes)):
            axes[idx].set_visible(False)

        plt.suptitle('Распределение значений по сенсорным каналам',
                    fontsize=14, fontweight='bold')
        plt.tight_layout()
        plt.savefig('02_distributions.png', dpi=150, bbox_inches='tight')
        plt.show()

    def analyze_missing_and_outliers(self):
        """Анализ пропусков и выбросов"""
        print("\n" + "=" * 70)
        print("ЭТАП 4: АНАЛИЗ ПРОПУСКОВ И ВЫБРОСОВ")
        print("=" * 70)

        if self.data is None:
            print("Нет данных для анализа")
            return

        channels = ['Accelerometer1RMS', 'Accelerometer2RMS', 'Current',
                   'Pressure', 'Temperature', 'Thermocouple', 'Voltage', 'Volume Flow RateRMS']
        existing_channels = [ch for ch in channels if ch in self.data.columns]

        # Анализ пропусков
        print("\nАнализ пропущенных значений:")
        missing_found = False
        for col in existing_channels:
            missing_count = self.data[col].isna().sum()
            missing_pct = missing_count / len(self.data) * 100
            if missing_count > 0:
                missing_found = True
            print(f"{col}: {missing_count} пропусков ({missing_pct:.4f}%)")

        if not missing_found:
            print("Пропуски отсутствуют во всех каналах")

        # Анализ выбросов по правилу 3 сигм
        print("\nАнализ выбросов (правило 3σ):")

        fig, axes = plt.subplots(2, 4, figsize=(16, 10))
        axes = axes.flatten()

        outliers_summary = {}

        for idx, col in enumerate(existing_channels):
            mean_val = self.data[col].mean()
            std_val = self.data[col].std()
            lower_bound = mean_val - 3 * std_val
            upper_bound = mean_val + 3 * std_val

            outliers = self.data[(self.data[col] < lower_bound) | (self.data[col] > upper_bound)]
            outlier_count = len(outliers)
            outlier_pct = outlier_count / len(self.data) * 100

            outliers_summary[col] = {
                'count': outlier_count,
                'percentage': outlier_pct,
                'lower_bound': lower_bound,
                'upper_bound': upper_bound
            }

            print(f"{col}: {outlier_count} выбросов ({outlier_pct:.2f}%)")

            # Box plot
            bp = axes[idx].boxplot(self.data[col].dropna(), vert=True, patch_artist=True)
            bp['boxes'][0].set_facecolor('lightblue')
            axes[idx].set_title(f'{col}', fontsize=10, fontweight='bold')
            axes[idx].set_ylabel('Значение')
            axes[idx].grid(True, alpha=0.3)

        for idx in range(len(existing_channels), len(axes)):
            axes[idx].set_visible(False)

        plt.suptitle('Диаграммы размаха для выявления выбросов',
                    fontsize=14, fontweight='bold')
        plt.tight_layout()
        plt.savefig('03_boxplots.png', dpi=150, bbox_inches='tight')
        plt.show()

    def analyze_ranges(self):
        """Анализ диапазонов значений"""
        print("\n" + "=" * 70)
        print("ЭТАП 5: АНАЛИЗ ДИАПАЗОНОВ ЗНАЧЕНИЙ")
        print("=" * 70)

        if self.data is None:
            print("Нет данных для анализа")
            return

        channels = ['Accelerometer1RMS', 'Accelerometer2RMS', 'Current',
                   'Pressure', 'Temperature', 'Thermocouple', 'Voltage', 'Volume Flow RateRMS']
        existing_channels = [ch for ch in channels if ch in self.data.columns]

        fig, ax = plt.subplots(figsize=(14, 6))

        # Семплирование для box plot (если данных много)
        data_to_plot = []
        for col in existing_channels:
            if len(self.data) > 50000:
                data_to_plot.append(self.data[col].sample(50000).values)
            else:
                data_to_plot.append(self.data[col].values)

        bp = ax.boxplot(data_to_plot, labels=existing_channels, patch_artist=True)

        for patch in bp['boxes']:
            patch.set_facecolor('lightgreen')
            patch.set_alpha(0.7)

        ax.set_ylabel('Значение', fontsize=11)
        ax.set_title('Сравнение диапазонов значений по каналам', fontsize=13, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='y')
        plt.xticks(rotation=45, ha='right')

        print("\nДиапазоны значений по каналам:")
        ranges = {}
        for col in existing_channels:
            min_val = self.data[col].min()
            max_val = self.data[col].max()
            ranges[col] = max_val - min_val
            print(f"{col}: min={min_val:.4f}, max={max_val:.4f}, размах={ranges[col]:.4f}")

        plt.tight_layout()
        plt.savefig('04_ranges_comparison.png', dpi=150, bbox_inches='tight')
        plt.show()

    def correlation_analysis(self):
        """Кореляционный анализ"""
        print("\n" + "=" * 70)
        print("ЭТАП 6: КОРРЕЛЯЦИОННЫЙ АНАЛИЗ")
        print("=" * 70)

        if self.data is None:
            print("  Нет данных для анализа")
            return

        channels = ['Accelerometer1RMS', 'Accelerometer2RMS', 'Current',
                   'Pressure', 'Temperature', 'Thermocouple', 'Voltage', 'Volume Flow RateRMS']
        existing_channels = [ch for ch in channels if ch in self.data.columns]

        # Семплирование для корреляции при большом объеме данных
        if len(self.data) > 50000:
            corr_data = self.data[existing_channels].sample(50000)
            print(f"Семплирование для корреляции: {len(self.data):,} → 50,000 точек")
        else:
            corr_data = self.data[existing_channels]

        corr_matrix = corr_data.corr()

        print("\nМатрица корреляции Пирсона:")
        print(corr_matrix.round(4).to_string())

        # Тепловая карта
        fig, ax = plt.subplots(figsize=(10, 8))

        sns.heatmap(corr_matrix, annot=True, fmt='.3f', cmap='RdBu_r',
                   center=0, square=True, linewidths=1,
                   cbar_kws={"shrink": 0.8}, ax=ax,
                   vmin=-1, vmax=1)

        ax.set_title('Тепловая карта корреляции между сенсорными каналами',
                    fontsize=14, fontweight='bold', pad=20)

        plt.tight_layout()
        plt.savefig('05_correlation_heatmap.png', dpi=150, bbox_inches='tight')
        plt.show()

        # Вывод сильных корреляций
        print("\nНаиболее сильные корреляции:")
        strong_corrs = []
        for i in range(len(corr_matrix.columns)):
            for j in range(i+1, len(corr_matrix.columns)):
                corr_val = corr_matrix.iloc[i, j]
                if abs(corr_val) > 0.5:
                    strong_corrs.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_val))

        strong_corrs.sort(key=lambda x: abs(x[2]), reverse=True)
        for col1, col2, val in strong_corrs[:5]:
            print(f"{col1} ↔ {col2}: {val:.3f}")

    def noise_analysis(self):
        """Поиск и анализ шумов"""
        print("\n" + "=" * 70)
        print("ЭТАП 7: ПОИСК И АНАЛИЗ ШУМОВ")
        print("=" * 70)

        if self.data is None:
            print("  Нет данных для анализа")
            return

        # Выбор ключевого канала (температура)
        key_channel = 'Temperature'
        if key_channel not in self.data.columns:
            key_channel = 'Accelerometer1RMS'

        print(f"Анализ шума для ключевого канала: {key_channel}")

        # Семплирование для декомпозиции
        sample_size = min(5000, len(self.data))
        data_sample = self.data[key_channel].iloc[:sample_size].copy()
        data_sample.index = pd.date_range(start='2024-01-01', periods=len(data_sample), freq='S')

        print(f"Используется {sample_size} точек для декомпозиции (из {len(self.data):,})")

        try:
            # Декомпозиция
            period = min(100, len(data_sample) // 10)
            if period < 2:
                period = 2
            result = seasonal_decompose(data_sample, model='additive', period=period)

            fig, axes = plt.subplots(4, 1, figsize=(14, 12), sharex=True)

            # Исходный ряд
            axes[0].plot(data_sample.index, result.observed, color='blue', linewidth=0.8)
            axes[0].set_ylabel('Значение', fontsize=10)
            axes[0].set_title('Исходный ряд', fontsize=11, fontweight='bold')
            axes[0].grid(True, alpha=0.3)

            # Тренд
            axes[1].plot(data_sample.index, result.trend, color='green', linewidth=0.8)
            axes[1].set_ylabel('Значение', fontsize=10)
            axes[1].set_title('Тренд', fontsize=11, fontweight='bold')
            axes[1].grid(True, alpha=0.3)

            # Сезонность
            axes[2].plot(data_sample.index, result.seasonal, color='orange', linewidth=0.8)
            axes[2].set_ylabel('Значение', fontsize=10)
            axes[2].set_title('Сезонная компонента', fontsize=11, fontweight='bold')
            axes[2].grid(True, alpha=0.3)

            # Остатки (шум)
            axes[3].plot(data_sample.index, result.resid, color='red', linewidth=0.8)
            axes[3].set_ylabel('Значение', fontsize=10)
            axes[3].set_title('Остатки (шум)', fontsize=11, fontweight='bold')
            axes[3].set_xlabel('Время', fontsize=10)
            axes[3].grid(True, alpha=0.3)

            plt.suptitle(f'Декомпозиция временного ряда: {key_channel}',
                        fontsize=14, fontweight='bold', y=1.02)
            plt.tight_layout()
            plt.savefig('06_decomposition.png', dpi=150, bbox_inches='tight')
            plt.show()

            # Расчет SNR
            signal_power = np.var(result.trend[~np.isnan(result.trend)]) + \
                          np.var(result.seasonal[~np.isnan(result.seasonal)])
            noise_power = np.var(result.resid[~np.isnan(result.resid)])
            snr = 10 * np.log10(signal_power / noise_power) if noise_power > 0 else float('inf')

            print(f"\nРасчет отношения сигнал/шум (SNR):")
            print(f"Мощность сигнала: {signal_power:.4f}")
            print(f"Мощность шума: {noise_power:.4f}")
            print(f"SNR = {snr:.2f} дБ")

            # Интерпретация SNR
            if snr > 20:
                snr_quality = "Отлично (>20 дБ)"
                snr_meaning = "Шум практически незаметен, данные очень чистые"
            elif snr > 10:
                snr_quality = "Хорошо (10-20 дБ)"
                snr_meaning = "Шум присутствует, но сигнал доминирует"
            elif snr > 0:
                snr_quality = "Удовлетворительно (0-10 дБ)"
                snr_meaning = "Сигнал и шум сравнимы по мощности, требуется фильтрация"
            else:
                snr_quality = "Плохо (<0 дБ)"
                snr_meaning = "Шум сильнее полезного сигнала"

            print(f"Качество: {snr_quality}")
            print(f"Интерпретация: {snr_meaning}")

            # Гистограмма остатков
            fig, ax = plt.subplots(figsize=(12, 5))

            resid_clean = result.resid.dropna()

            ax.hist(resid_clean, bins=50, density=True, alpha=0.7,
                   edgecolor='black', color='red', label='Фактическое распределение')

            # Нормальное распределение для сравнения
            mu, std = norm.fit(resid_clean)
            xmin, xmax = ax.get_xlim()
            x = np.linspace(xmin, xmax, 100)
            ax.plot(x, norm.pdf(x, mu, std), 'k--', linewidth=2,
                   label=f'Нормальное распределение\n(μ={mu:.2f}, σ={std:.2f})')

            ax.set_xlabel('Значение остатков', fontsize=11)
            ax.set_ylabel('Плотность вероятности', fontsize=11)
            ax.set_title('Распределение шума (остатков) временного ряда',
                        fontsize=13, fontweight='bold')
            ax.legend()
            ax.grid(True, alpha=0.3)

            plt.tight_layout()
            plt.savefig('07_noise_distribution.png', dpi=150, bbox_inches='tight')
            plt.show()

            # Анализ формы распределения
            skewness = skew(resid_clean)
            kurt_val = kurtosis(resid_clean)

            print(f"\nХарактеристики распределения шума:")
            print(f"Асимметрия (skewness): {skewness:.3f}")
            print(f"Эксцесс (kurtosis): {kurt_val:.3f}")

            if abs(skewness) < 0.5 and abs(kurt_val) < 1:
                shape_desc = "Близко к нормальному (колоколообразное, симметричное)"
                shape_meaning = "Шум случаен, близок к классическому «белому шуму»"
                recommendation = "Фильтрация не требуется"
            elif abs(skewness) > 1:
                shape_desc = "Асимметричное распределение"
                shape_meaning = "Возможна систематическая ошибка измерений"
                recommendation = "Рекомендуется логарифмическое преобразование"
            elif kurt_val > 3:
                shape_desc = "Тяжелые хвосты (эксцесс > 3)"
                shape_meaning = "Присутствуют редкие, но сильные выбросы"
                recommendation = "Рекомендуется медианный фильтр"
            else:
                shape_desc = "Умеренное отклонение от нормального"
                shape_meaning = "Шум имеет регулярную структуру"
                recommendation = "Скользящее среднее для сглаживания"

            print(f"Форма распределения: {shape_desc}")
            print(f"Интерпретация: {shape_meaning}")
            print(f"Рекомендация по фильтрации: {recommendation}")

        except Exception as e:
            print(f"Декомпозиция не выполнена: {e}")

    def detect_anomalies_analysis(self):
        """ДОПОЛНИТЕЛЬНЫЙ АНАЛИЗ: Обнаружение аномалий"""
        print("\n" + "=" * 70)
        print("ДОПОЛНИТЕЛЬНЫЙ АНАЛИЗ: ОБНАРУЖЕНИЕ АНОМАЛИЙ")
        print("=" * 70)

        if self.data is None:
            print("Нет данных для анализа")
            return

        if 'anomaly' not in self.data.columns:
            print("В данных отсутствует разметка аномалий")
            return

        # Общая статистика
        total_anomalies = self.data['anomaly'].sum()
        anomaly_pct = total_anomalies / len(self.data) * 100

        print(f"\nРезультаты анализа размеченных аномалий:")
        print(f"Всего аномалий: {total_anomalies} ({anomaly_pct:.2f}% от всех данных)")

        if 'changepoint' in self.data.columns:
            total_changepoints = self.data['changepoint'].sum()
            print(f"Точек смены состояния (changepoint): {total_changepoints}")

        # Анализ по файлам
        print("\nАнализ аномалий по отдельным файлам:")
        for info in self.file_info:
            print(f"{info['file_name']}: {info['anomaly_count']} аномалий")

        # Визуализация аномалий на нескольких каналах
        fig, axes = plt.subplots(3, 1, figsize=(15, 10), sharex=True)

        # Даунсемплинг для визуализации
        total_points = len(self.data)
        max_viz_points = 50000
        if total_points > max_viz_points:
            step = total_points // max_viz_points
            viz_data = self.data.iloc[::step].copy()
            viz_anomalies = self.data['anomaly'].iloc[::step]
            print(f"Даунсемплинг для визуализации: {total_points:,} → {len(viz_data):,} точек")
        else:
            viz_data = self.data
            viz_anomalies = self.data['anomaly']

        # График температуры
        if 'Temperature' in viz_data.columns:
            axes[0].plot(viz_data['timestamp'], viz_data['Temperature'],
                        'b-', linewidth=0.5, alpha=0.7, label='Температура', rasterized=True)
            anomaly_mask = viz_anomalies == 1
            if anomaly_mask.any():
                axes[0].scatter(viz_data[anomaly_mask]['timestamp'],
                               viz_data[anomaly_mask]['Temperature'],
                               color='red', s=20, alpha=0.8, label='Аномалии', zorder=5)
            axes[0].set_ylabel('Температура (°C)')
            axes[0].set_title('Температура с выделением аномалий')
            axes[0].legend(loc='upper left')
            axes[0].grid(True, alpha=0.3)

        # График вибрации
        if 'Accelerometer1RMS' in viz_data.columns:
            axes[1].plot(viz_data['timestamp'], viz_data['Accelerometer1RMS'],
                        'g-', linewidth=0.5, alpha=0.7, label='Вибрация X', rasterized=True)
            if anomaly_mask.any():
                axes[1].scatter(viz_data[anomaly_mask]['timestamp'],
                               viz_data[anomaly_mask]['Accelerometer1RMS'],
                               color='red', s=20, alpha=0.8, label='Аномалии', zorder=5)
            axes[1].set_ylabel('Амплитуда вибрации')
            axes[1].set_title('Вибрация (ось X) с выделением аномалий')
            axes[1].legend(loc='upper left')
            axes[1].grid(True, alpha=0.3)

        # График тока
        if 'Current' in viz_data.columns:
            axes[2].plot(viz_data['timestamp'], viz_data['Current'],
                        'purple', linewidth=0.5, alpha=0.7, label='Ток', rasterized=True)
            if anomaly_mask.any():
                axes[2].scatter(viz_data[anomaly_mask]['timestamp'],
                               viz_data[anomaly_mask]['Current'],
                               color='red', s=20, alpha=0.8, label='Аномалии', zorder=5)
            axes[2].set_xlabel('Временные точки')
            axes[2].set_ylabel('Ток (А)')
            axes[2].set_title('Ток с выделением аномалий')
            axes[2].legend(loc='upper left')
            axes[2].grid(True, alpha=0.3)

        plt.suptitle('Обнаружение аномалий в данных SKAB',
                    fontsize=14, fontweight='bold', y=1.02)
        plt.tight_layout()
        plt.savefig('08_anomaly_detection.png', dpi=150, bbox_inches='tight')
        plt.show()

    def run_full_analysis(self):
        """Запуск полного анализа в соответствии с этапами УММ."""
        print("\n" + "=" * 70)
        print("НАЧАЛО ПОЛНОГО АНАЛИЗА ВРЕМЕННЫХ РЯДОВ (SKAB)")
        print("=" * 70)

        self.load_all_data()

        if self.data is not None and len(self.data) > 0:
            self.visualize_initial_data()
            self.statistical_analysis()
            self.analyze_missing_and_outliers()
            self.analyze_ranges()
            self.correlation_analysis()
            self.noise_analysis()
            self.detect_anomalies_analysis()

def main():
    data_path = "/content/value1"

    # Проверка наличия файлов
    csv_files = [f for f in os.listdir(data_path) if f.endswith('.csv')]
    if len(csv_files) == 0:
        # Проверка в подпапке valve
        valve_path = os.path.join(data_path, "valve1")
        if os.path.exists(valve_path):
            data_path = valve_path
            csv_files = [f for f in os.listdir(data_path) if f.endswith('.csv')]

    print(f"Найдено CSV-файлов для анализа: {len(csv_files)}")
    if len(csv_files) > 0:
        print("Файлы:", ", ".join(sorted(csv_files)[:5]),
              "..." if len(csv_files) > 5 else "")

    analyzer = SKABAnomalyAnalyzer(data_path=data_path, critical_temp=85)
    analyzer.run_full_analysis()


if __name__ == "__main__":
    main()
```
