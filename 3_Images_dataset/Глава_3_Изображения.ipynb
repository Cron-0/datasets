```python
import os
import glob
import xml.etree.ElementTree as ET
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter, defaultdict
from PIL import Image
import warnings
warnings.filterwarnings('ignore')

# СЛОВАРЬ ПЕРЕВОДА ФРАНЦУЗСКИХ НАЗВАНИЙ НА РУССКИЕ
TRANSLATION = {
    # Типы документов
    'Lettre manuscrite': 'Рукописное письмо',
    'Lettre dactylographiée': 'Машинописное письмо',
    'Questionnaire': 'Анкета',

    # Типы полей (bounding boxes)
    'Coordonnées Expéditeur': 'Корреспондент (Отправитель)',
    'Coordonnées Destinataire': 'Адресат (Получатель)',
    'Date, Lieu': 'Дата и место',
    'Signature': 'Подписант (Подпись)',
    'Objet': 'Тема письма',
    'Corps de texte': 'Тело письма',
    'Ouverture': 'Обращение',
    'Référence client': 'Номер письма (Ref. клиента)',
    'Code d\'envoi': 'Код отправки',
    'Nom': 'Фамилия',
    'Prénom': 'Имя',
    'Adresse': 'Адрес',
    'Code Postal': 'Почтовый индекс',
    'Ville': 'Город',
    'Téléphone': 'Телефон',
    'Catégorie Scénario': 'Категория сценария',
    'Objet Scénario': 'Содержание сценария',
    'Interlocuteur': 'Контрагент',
    'Adresse Interlocuteur': 'Адрес контрагента',
    'PS/PJ':'Послесловие/Приложение',

    # Сценарии
    'Relance de courrier sans réponse': 'Безответное почтовое письмо',
    'Difficulté de paiement': 'Трудности с денежным переводом',
    'Changement de données personnelles': 'Изменение персональных данных',
    'Demande d\'information': 'Запрос информации',
    'Modification de contrat / Commande': 'Изменение контракта / заказа',
    'Fermeture de compte': 'Закрытие счёта',
    'Ouverture de compte': 'Открытие счёта',
    'Gestion de sinistre': 'Управление страховым случаем',
    'Demande de documentation': 'Запрос документации',
    'Réclamation': 'Претензия',
    'Information': 'Информационное письмо',
}

def translate(text):
    """Перевод французского текста на русский"""
    if not text:
        return text
    if text in TRANSLATION:
        return TRANSLATION[text]
    return f"{text} (фр.)"

# Загрузка и первичное знакомство с данными
DATA_PATH = "/content/drive/MyDrive/DVD1_TIF"

# Поиск всех XML-файлов (аннотации писем)
xml_files = glob.glob(os.path.join(DATA_PATH, "*_L.xml"))
print(f"Найдено XML-аннотаций: {len(xml_files)}")

# Поиск всех изображений писем
image_files = glob.glob(os.path.join(DATA_PATH, "*_L.tif"))
print(f"Найдено изображений писем: {len(image_files)}")

# Проверка соответствия
assert len(xml_files) == len(image_files), "Несоответствие количества файлов"
print(f"Всего пар (изображение + аннотация): {len(xml_files)}")

# Функция для парсинга XML аннотации
def parse_rimes_xml(xml_path):
    """Извлечение всех полей из XML аннотаций RIMES"""
    tree = ET.parse(xml_path)
    root = tree.getroot()

    data = {
        'document_type': translate(root.findtext('document', '')),
        'scenario': translate(root.findtext('scenario', '')),
        'writer': root.findtext('writer', ''),
        'sender_person': root.findtext('sender_person', ''),
        'sender_company': root.findtext('sender_company', ''),
        'receiver_person': root.findtext('receiver_person', ''),
        'receiver_company': root.findtext('receiver_company', ''),
        'subject': root.findtext('subject', ''),
        'date': root.findtext('date', ''),
        'place': root.findtext('place', ''),
        'boxes': []
    }

    for box in root.findall('box'):
        box_type = box.findtext('type', '')
        box_data = {
            'type': translate(box_type),
            'type_original': box_type,
            'text': box.findtext('text', ''),
            'top_left_x': int(box.get('top_left_x', 0)),
            'top_left_y': int(box.get('top_left_y', 0)),
            'bottom_right_x': int(box.get('bottom_right_x', 0)),
            'bottom_right_y': int(box.get('bottom_right_y', 0)),
        }
        box_data['width'] = box_data['bottom_right_x'] - box_data['top_left_x']
        box_data['height'] = box_data['bottom_right_y'] - box_data['top_left_y']
        data['boxes'].append(box_data)

    return data

# Парсинг всех XML файлов
all_annotations = []
for xml_file in xml_files:
    try:
        ann = parse_rimes_xml(xml_file)
        ann['file_name'] = os.path.basename(xml_file).replace('_L.xml', '_L.tif')
        all_annotations.append(ann)
    except Exception as e:
        print(f"Ошибка при парсинге {xml_file}: {e}")

print(f"Успешно загружено {len(all_annotations)} аннотаций")

print("\n" + "=" * 80)
print("РЕЗУЛЬТАТЫ АНАЛИЗА ДЕЛОВОЙ КОРРЕСПОНДЕНЦИИ")
print("=" * 80)

# Сбор данных по 5 обязательным полям для каждого документа
required_fields_data = []

for ann in all_annotations:
    doc_data = {
        'file_name': ann['file_name'],
        'Корреспондент (Отправитель)': '',
        'Дата письма': '',
        'Номер письма (Reference)': '',
        'Адресат (Получатель)': '',
        'Подписант (Подпись)': '',
        'Доп. информация': ''
    }

    if ann['sender_person']:
        doc_data['Корреспондент (Отправитель)'] = ann['sender_person']
    elif ann['sender_company']:
        doc_data['Корреспондент (Отправитель)'] = ann['sender_company']

    if ann['date']:
        doc_data['Дата письма'] = f"{ann['date']} ({ann['place']})" if ann['place'] else ann['date']

    if ann['receiver_person']:
        doc_data['Адресат (Получатель)'] = ann['receiver_person']
    elif ann['receiver_company']:
        doc_data['Адресат (Получатель)'] = ann['receiver_company']

    for box in ann['boxes']:
        if 'Корреспондент' in box['type'] and not doc_data['Корреспондент (Отправитель)']:
            doc_data['Корреспондент (Отправитель)'] = box['text'].replace('\n', ', ')
        elif 'Адресат' in box['type'] and not doc_data['Адресат (Получатель)']:
            doc_data['Адресат (Получатель)'] = box['text'].replace('\n', ', ')
        elif 'Дата' in box['type'] and not doc_data['Дата письма']:
            doc_data['Дата письма'] = box['text']
        elif 'Подписант' in box['type'] or 'Signature' in box['type_original']:
            doc_data['Подписант (Подпись)'] = box['text'] if box['text'] else '[Рукописная подпись, текст не распознан]'
        elif 'Référence' in box['type_original'] or 'reference' in box['type_original'].lower():
            doc_data['Номер письма (Reference)'] = box['text']

    if not doc_data['Номер письма (Reference)']:
        doc_data['Номер письма (Reference)'] = ann['file_name'].replace('_L.tif', '')

    doc_data['Доп. информация'] = f"Сценарий: {ann['scenario']}, Автор: {ann['writer']}"
    required_fields_data.append(doc_data)

df_required = pd.DataFrame(required_fields_data)

print("\n" + "=" * 80)
print("ТАБЛИЦА 1: ОСНОВНЫЕ ПОЛЯ ДЕЛОВОЙ КОРРЕСПОНДЕНЦИИ (первые 10 записей)")
print("=" * 80)
display(df_required.head(10))

print("\n" + "=" * 80)
print("СТАТИСТИКА ЗАПОЛНЕННОСТИ ОБЯЗАТЕЛЬНЫХ ПОЛЕЙ")
print("=" * 80)
for field in ['Корреспондент (Отправитель)', 'Дата письма', 'Номер письма (Reference)',
              'Адресат (Получатель)', 'Подписант (Подпись)']:
    non_empty = df_required[field].astype(str).str.len() > 0
    non_empty_count = non_empty.sum()
    print(f"  {field}: {non_empty_count}/{len(df_required)} ({non_empty_count/len(df_required)*100:.1f}%)")

# Создание DataFrame для метаданных
df_meta = pd.DataFrame([{
    'file_name': ann['file_name'],
    'Тип документа': ann['document_type'],
    'Сценарий': ann['scenario'],
    'Автор (почерк)': ann['writer'],
    'Отправитель': ann['sender_person'] if ann['sender_person'] else ann['sender_company'],
    'Получатель': ann['receiver_person'] if ann['receiver_person'] else ann['receiver_company'],
    'Дата': ann['date'],
    'Место': ann['place'],
    'Тема': ann['subject'],
    'Количество полей': len(ann['boxes'])
} for ann in all_annotations])

print("\n" + "=" * 80)
print("ТАБЛИЦА 2: МЕТАДАННЫЕ ДОКУМЕНТОВ (первые 10 записей)")
print("=" * 80)
display(df_meta.head(10))

# 1. Анализ количества и баланса классов
print("\n" + "=" * 50)
print("СТАТИСТИКА ПО ТИПАМ ДОКУМЕНТОВ")
print("=" * 50)
doc_type_counts = df_meta['Тип документа'].value_counts()
for doc_type, count in doc_type_counts.items():
    print(f"  {doc_type}: {count} ({count/len(df_meta)*100:.1f}%)")

print("\n" + "=" * 50)
print("СТАТИСТИКА ПО СЦЕНАРИЯМ (ЗАДАЧАМ)")
print("=" * 50)
scenario_counts = df_meta['Сценарий'].value_counts()
for scenario, count in scenario_counts.items():
    print(f"  {scenario}: {count} ({count/len(df_meta)*100:.1f}%)")

print("\n" + "=" * 50)
print("СТАТИСТИКА ПО АВТОРАМ (ПОЧЕРКАМ)")
print("=" * 50)
writer_counts = df_meta['Автор (почерк)'].value_counts()
for writer, count in writer_counts.head(10).items():
    print(f"  {writer}: {count}")

# Визуализация распределений
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].pie(doc_type_counts.values, labels=doc_type_counts.index, autopct='%1.1f%%')
axes[0].set_title('Распределение по типам документов', fontsize=12, fontweight='bold')

top_scenarios = scenario_counts.head(6)
axes[1].barh(range(len(top_scenarios)), top_scenarios.values, color='skyblue')
axes[1].set_yticks(range(len(top_scenarios)))
axes[1].set_yticklabels(top_scenarios.index)
axes[1].set_xlabel('Количество')
axes[1].set_title('Топ-6 сценариев', fontsize=12, fontweight='bold')

axes[2].hist(df_meta['Количество полей'], bins=20, color='lightgreen', edgecolor='black')
axes[2].set_xlabel('Количество структурных полей')
axes[2].set_ylabel('Частота')
axes[2].set_title('Распределение числа полей на документ', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

# 2. Примеры типичных изображений
print("\n" + "=" * 50)
print("ПРИМЕРЫ ТИПИЧНЫХ ИЗОБРАЖЕНИЙ")
print("=" * 50)

def show_sample_images(image_files, num_samples=3):
    """Отображение примеров изображений"""
    fig, axes = plt.subplots(1, num_samples, figsize=(5*num_samples, 7))
    if num_samples == 1:
        axes = [axes]

    for i in range(min(num_samples, len(image_files))):
        img_path = image_files[i]
        img = Image.open(img_path)
        axes[i].imshow(img, cmap='gray')
        axes[i].set_title(f"Пример {i+1}: {os.path.basename(img_path)}", fontsize=10)
        axes[i].axis('off')

    plt.tight_layout()
    plt.savefig('sample_images.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("Рисунок. Примеры типичных изображений из набора данных")

show_sample_images(image_files, num_samples=3)

# 3. Оценка качества изображений
print("\n" + "=" * 50)
print("ОЦЕНКА КАЧЕСТВА ИЗОБРАЖЕНИЙ")
print("=" * 50)

def analyze_image_quality(image_path):
    """Анализ качества изображения"""
    try:
        img = Image.open(image_path)
        img_array = np.array(img)
        height, width = img_array.shape[:2]

        # Контрастность (стандартное отклонение)
        if len(img_array.shape) == 2:
            contrast = np.std(img_array)
            brightness = np.mean(img_array)
        else:
            contrast = np.std(np.mean(img_array, axis=2))
            brightness = np.mean(np.mean(img_array, axis=2))

        return {
            'width': width,
            'height': height,
            'resolution': width * height,
            'aspect_ratio': width / height,
            'contrast': contrast,
            'brightness': brightness
        }
    except Exception as e:
        print(f"Ошибка при анализе {image_path}: {e}")
        return None

# Анализ первых 100 изображений (для скорости, можно увеличить)
sample_size = min(100, len(image_files))
quality_stats = []
for img_path in image_files[:sample_size]:
    stats = analyze_image_quality(img_path)
    if stats:
        quality_stats.append(stats)

df_quality = pd.DataFrame(quality_stats)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Разрешение
axes[0, 0].hist(df_quality['resolution'] / 1e6, bins=20, color='coral', edgecolor='black')
axes[0, 0].axvline(df_quality['resolution'].mean() / 1e6, color='red', linestyle='--',
                   label=f"Среднее: {df_quality['resolution'].mean()/1e6:.1f} МП")
axes[0, 0].set_xlabel('Разрешение (мегапиксели)')
axes[0, 0].set_ylabel('Количество')
axes[0, 0].set_title('Распределение разрешения', fontsize=12, fontweight='bold')
axes[0, 0].legend()

# Соотношение сторон
axes[0, 1].hist(df_quality['aspect_ratio'], bins=20, color='lightblue', edgecolor='black')
axes[0, 1].set_xlabel('Соотношение сторон (ширина/высота)')
axes[0, 1].set_ylabel('Количество')
axes[0, 1].set_title('Распределение соотношения сторон', fontsize=12, fontweight='bold')

# Контрастность
axes[1, 0].hist(df_quality['contrast'], bins=20, color='lightgreen', edgecolor='black')
axes[1, 0].set_xlabel('Контрастность (стандартное отклонение)')
axes[1, 0].set_ylabel('Количество')
axes[1, 0].set_title('Распределение контрастности', fontsize=12, fontweight='bold')

# Яркость
axes[1, 1].hist(df_quality['brightness'], bins=20, color='plum', edgecolor='black')
axes[1, 1].set_xlabel('Яркость')
axes[1, 1].set_ylabel('Количество')
axes[1, 1].set_title('Распределение яркости', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('image_quality.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\nСтатистика качества изображений (на основе {sample_size} файлов):")
print(f"  Среднее разрешение: {df_quality['resolution'].mean() / 1e6:.2f} МП")
print(f"  Средний размер: {df_quality['width'].mean():.0f}×{df_quality['height'].mean():.0f}")
print(f"  Среднее соотношение сторон: {df_quality['aspect_ratio'].mean():.3f}")
print(f"  Средняя контрастность: {df_quality['contrast'].mean():.2f}")
print(f"  Средняя яркость: {df_quality['brightness'].mean():.2f}")

# 4. Анализ типов аннотаций (bounding boxes)
all_box_types = []
for ann in all_annotations:
    for box in ann['boxes']:
        all_box_types.append(box['type'])

box_type_counts = Counter(all_box_types)
print("\n" + "=" * 50)
print("ТИПЫ СТРУКТУРНЫХ ПОЛЕЙ (BOUNDING BOXES)")
print("=" * 50)
for box_type, count in sorted(box_type_counts.items(), key=lambda x: -x[1]):
    print(f"  {box_type}: {count}")

# Визуализация типов полей
plt.figure(figsize=(12, 8))
colors = plt.cm.Set3(range(len(box_type_counts)))
plt.pie(box_type_counts.values(), labels=box_type_counts.keys(), autopct='%1.1f%%', colors=colors)
plt.title('Распределение типов структурных элементов', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('box_types.png', dpi=150, bbox_inches='tight')
plt.show()

# 5. Анализ длин текстов в разных типах полей
field_text_lengths = defaultdict(list)
for ann in all_annotations:
    for box in ann['boxes']:
        if box['text']:
            field_text_lengths[box['type']].append(len(box['text']))

fig, ax = plt.subplots(figsize=(12, 6))
data_to_plot = []
labels = []
for field_type in ['Корреспондент (Отправитель)', 'Адресат (Получатель)', 'Тело письма',
                   'Тема письма', 'Дата и место', 'Подписант (Подпись)']:
    if field_type in field_text_lengths:
        data_to_plot.append(field_text_lengths[field_type])
        labels.append(field_type)

bp = ax.boxplot(data_to_plot, labels=labels, patch_artist=True)
for patch in bp['boxes']:
    patch.set_facecolor('lightblue')

ax.set_ylabel('Длина текста (символы)')
ax.set_title('Распределение длины текста по типам полей', fontsize=12, fontweight='bold')
ax.set_yscale('log')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('text_lengths.png', dpi=150, bbox_inches='tight')
plt.show()

# 6. Анализ наличия критических полей
critical_fields_ru = ['Корреспондент (Отправитель)', 'Адресат (Получатель)', 'Дата и место']
field_presence = {field: 0 for field in critical_fields_ru}

for ann in all_annotations:
    present_fields = [box['type'] for box in ann['boxes']]
    for field in critical_fields_ru:
        if field in present_fields:
            field_presence[field] += 1

print("\n" + "=" * 50)
print("НАЛИЧИЕ КЛЮЧЕВЫХ ПОЛЕЙ В ДОКУМЕНТАХ")
print("=" * 50)
for field, count in field_presence.items():
    print(f"  {field}: {count}/{len(all_annotations)} ({count/len(all_annotations)*100:.1f}%)")

# 7. Пример разметки документа
sample_ann = all_annotations[0]
print("\n" + "=" * 50)
print("ПРИМЕР РАЗМЕТКИ ДОКУМЕНТА (с переводом)")
print("=" * 50)
print(f"Файл: {sample_ann['file_name']}")
print(f"Автор: {sample_ann['writer']}")
print(f"Сценарий: {sample_ann['scenario']}")
print("\nСтруктурные элементы:")
for box in sample_ann['boxes']:
    if box['type'] in critical_fields_ru + ['Подписант (Подпись)', 'Тема письма']:
        print(f"\n  [{box['type']}]")
        print(f"    Координаты: ({box['top_left_x']}, {box['top_left_y']}) → ({box['bottom_right_x']}, {box['bottom_right_y']})")
        print(f"    Размер: {box['width']}×{box['height']} пикселей")
        if box['text']:
            preview = box['text'][:100].replace('\n', ' ')
            print(f"    Текст: {preview}...")

# 8. Сохранение результатов
df_required.to_csv('rimes_required_fields.csv', index=False, encoding='utf-8-sig')
df_meta.to_csv('rimes_metadata.csv', index=False, encoding='utf-8-sig')
box_type_df = pd.DataFrame(box_type_counts.items(), columns=['box_type', 'count'])
box_type_df.to_csv('box_type_distribution.csv', index=False, encoding='utf-8-sig')
```
